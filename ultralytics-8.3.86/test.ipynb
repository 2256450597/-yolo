```python
#导入YOLO模块
from ultralytics import YOLO

#初始化用于图像检测的YOLO模型,括号内写入模型的路径（相对或绝对路径）
model = YOLO("yolo11n.pt")

#把检测元作为参数给model
result = model(source="ultralytics/assets/bus.jpg", show=True)

#代码形式进行目标检测，此时不会将检测结果的图像保存。想保存show。摄像头source=0，（对应的设备号）show=True,实时看到结果

#目标检测的文字结果不是
```

```python
#查看det_result.boxes的类型和说明调用属性
det_boxes = det_result.boxes
```

```python
#获取模型可以识别的字典数据
det_dict = det_result.names
det_dict[0]
```

```python
#尝试对det_cls矩阵进行索引
det_cls[0]
```

```python
#将tensor转换为可以正常索引的格式。
det_cls_arr = det_cls.numpy()
```

```python
#将cls矩阵的数值作为访问字典的键值，就可以获取到具体的物体类型字符串了
det_dict[int(det_cls_arr[1])] 
```

```python
#统计cls矩阵中出现的不同物体类别以及每种类别的数量
import numpy as np
#第一个返回值是矩阵内出现的不同元素有哪些
#第二个返回值是每个不同元素出现的次数
det_cls_unique_arr,det_cls_num_arr = np.unique(det_cls_arr, return_counts=True)
```

```python
#使用for循环遍历cls矩阵，把所有物体的类别都print出来
size = int (len(det_cls_arr))
for i in range(size):
    cls_index = det_cls_arr[i]
    print(f'类别索引号为{det_cls_arr[i]}')
    #使用每一轮得到的索引号访问字典 得到检测的物体
    cls_obj = det_dict[cls_index]
    print(f'物体名称:[cls_obj]')
    #显示对应物体的数量
    cls_num = det_cls_num_arr[i]
    print(f'{cls_obj}物体数量:{cls_num}')
```

```python
#获取检测结果
result = model(source="ultralytics/assets/bus.jpg", verbose=False)
#取出列表中的检测对象
det_result = result[0]
#获取存放检测对象的字典
det_dict = det_result.names
#获取所有检测结果的综合
det_boxes = det_result.boxes
#或缺检测结果中的类别索引号矩阵
det_cls = det_boxes.cls
#将cls矩阵转化为能够正常提取的成员numpy矩阵
det_cls_arr = det_cls.numpy()
#使用for循环遍历cls矩阵，把所有物体的类别都print出来
size = int (len(det_cls_arr))
for i in range(size):
    cls_index = det_cls_arr[i]
    print(f'类别索引号为{det_cls_arr[i]}')
    #使用每一轮得到的索引号访问字典 得到检测的物体
    cls_obj = det_dict[cls_index]
    print(f'物体名称:[cls_obj]')
```

```python
import numpy as np
#获取检测结果
result = model(source="ultralytics/assets/bus.jpg", verbose=False)
#取出列表中的检测对象
det_result = result[0]
#获取存放检测对象的字典
det_dict = det_result.names
#获取所有检测结果的综合
det_boxes = det_result.boxes
#或缺检测结果中的类别索引号矩阵
det_cls = det_boxes.cls
#将cls矩阵转化为能够正常提取的成员numpy矩阵
det_cls_arr = det_cls.numpy()
det_cls_unique_arr,det_cls_num_arr = np.unique(det_cls_arr, return_counts=True)
#使用for循环遍历cls矩阵，把所有物体的类别都print出来
size = int (len(det_cls_arr))
size = int (len(det_cls_unique_arr))
for i in range(size):
    cls_index = det_cls_unique_arr[i]
    #print(f'类别索引号为{det_cls_unique_arr[i]}')
    #使用每一轮得到的索引号访问字典 得到检测的物体
    cls_obj = det_dict[cls_index]
    #print(f'物体名称:[cls_obj]')
    #显示对应物体的数量
    cls_num = det_cls_num_arr[i]
    print(f'{cls_obj}物体数量:{cls_num}')
```
